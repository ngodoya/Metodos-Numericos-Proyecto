```python
import sys
import os

sys.path.append(os.path.abspath(os.path.join("..")))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.models.logistic_regression import LogisticRegression
from src.optimizers.gradient_descent import GradientDescent
from src.optimizers.newton_raphson import NewtonRaphson

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
np.random.seed(42)  # Reproducibilidad

# Generamos 200 estudiantes
n_samples = 200

# Feature 1: Horas de estudio (1 a 10 hrs)
hours_studied = np.random.uniform(1, 10, n_samples)

# Feature 2: Promedio de calificaciones previas (50 a 100 pts)
previous_score = np.random.uniform(50, 100, n_samples)

# Construimos una combinación lineal real
logits = -8 + 0.8 * hours_studied + 0.08 * previous_score

# Convertimos a probabilidad real con la sigmoide
true_probs = 1 / (1 + np.exp(-logits))

# Generamos las etiquetas binarias (0 o 1) con una distribución Bernoulli
y = (np.random.binomial(1, true_probs)).astype(int)

# Unimos las características en una matriz X
X = np.column_stack((hours_studied, previous_score))

print(f"Dataset generado:")
print(f"- Muestras totales: {X.shape[0]}")
print(f"- Estudiantes Reprobados (0): {np.sum(y == 0)}")
print(f"- Estudiantes Aprobados (1): {np.sum(y == 1)}")
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=X[:, 0],
    y=X[:, 1],
    hue=y,
    palette={0: "red", 1: "green"},
    alpha=0.8,
    s=70,
)
plt.title("Dataset: Aprobación de Estudiantes")
plt.xlabel("Horas de Estudio")
plt.ylabel("Puntaje Previo")
plt.legend(title="Aprobado", labels=["No (0)", "Sí (1)"])
plt.show()
```

```python
# 1. Descenso del Gradiente
gd_optimizer = GradientDescent(learning_rate=0.01, max_iter=2000, tol=1e-6)
model_gd = LogisticRegression(optimizer=gd_optimizer)
model_gd.fit(X, y)

# 2. Newton-Raphson
nr_optimizer = NewtonRaphson(max_iter=50, tol=1e-6)
model_nr = LogisticRegression(optimizer=nr_optimizer)
model_nr.fit(X, y)

print("=== RESULTADOS DEL ENTRENAMIENTO ===")
print(f"Descenso del Gradiente -> Iteraciones: {len(gd_optimizer.history)}")
print(f"  - Intercepto (b): {model_gd.intercept_:.4f}")
print(f"  - Coeficientes (w): {model_gd.coef_}")

print(f"\nNewton-Raphson       -> Iteraciones: {len(nr_optimizer.history)}")
print(f"  - Intercepto (b): {model_nr.intercept_:.4f}")
print(f"  - Coeficientes (w): {model_nr.coef_}")
```

```python
import time
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LogisticRegression as SklearnLogistic
from sklearn.metrics import accuracy_score, confusion_matrix, log_loss

print("=" * 60)
print("BENCHMARKING: PyStatistics vs. Scikit-Learn")
print("=" * 60)

# -------------------------------------------------------------
# 1. PyStatistics - Descenso del Gradiente
# -------------------------------------------------------------
start_time = time.time()
gd_opt = GradientDescent(learning_rate=0.01, max_iter=2000, tol=1e-6)
pystat_gd = LogisticRegression(optimizer=gd_opt)
pystat_gd.fit(X, y)
time_pystat_gd = time.time() - start_time

preds_gd = pystat_gd.predict(X)
probs_gd = pystat_gd.predict_proba(X)

# -------------------------------------------------------------
# 2. PyStatistics - Newton-Raphson
# -------------------------------------------------------------
start_time = time.time()
nr_opt = NewtonRaphson(max_iter=50, tol=1e-6)
pystat_nr = LogisticRegression(optimizer=nr_opt)
pystat_nr.fit(X, y)
time_pystat_nr = time.time() - start_time

preds_nr = pystat_nr.predict(X)
probs_nr = pystat_nr.predict_proba(X)

# -------------------------------------------------------------
# 3. Scikit-Learn (Sin regularización: C=np.inf)
# -------------------------------------------------------------
start_time = time.time()
sklearn_model = SklearnLogistic(C=np.inf, solver="lbfgs", max_iter=1000)
sklearn_model.fit(X, y)
time_sklearn = time.time() - start_time

preds_sk = sklearn_model.predict(X)
probs_sk = sklearn_model.predict_proba(X)[:, 1]

# -------------------------------------------------------------
# TABLA DE RESULTADOS
# -------------------------------------------------------------
results = pd.DataFrame(
    {
        "Modelo / Optimizador": [
            "PyStat (Gradient Descent)",
            "PyStat (Newton-Raphson)",
            "Scikit-Learn (L-BFGS)",
        ],
        "Iteraciones": [
            len(gd_opt.history),
            len(nr_opt.history),
            sklearn_model.n_iter_[0],
        ],
        "Tiempo (s)": [
            f"{time_pystat_gd:.4f}",
            f"{time_pystat_nr:.4f}",
            f"{time_sklearn:.4f}",
        ],
        "Accuracy": [
            accuracy_score(y, preds_gd),
            accuracy_score(y, preds_nr),
            accuracy_score(y, preds_sk),
        ],
        "Log-Loss": [
            log_loss(y, probs_gd),
            log_loss(y, probs_nr),
            log_loss(y, probs_sk),
        ],
        "Intercepto ($b$)": [
            f"{pystat_gd.intercept_:.4f}",
            f"{pystat_nr.intercept_:.4f}",
            f"{sklearn_model.intercept_[0]:.4f}",
        ],
        "Coef ($w_1, w_2$)": [
            np.round(pystat_gd.coef_, 4),
            np.round(pystat_nr.coef_, 4),
            np.round(sklearn_model.coef_[0], 4),
        ],
    }
)

display(results)

# -------------------------------------------------------------
# VISUALIZACIÓN 1: MATRICES DE CONFUSIÓN
# -------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

cm_nr = confusion_matrix(y, preds_nr)
sns.heatmap(cm_nr, annot=True, fmt="d", cmap="Blues", ax=axes[0], cbar=False)
axes[0].set_title("PyStat (Newton-Raphson)")
axes[0].set_xlabel("Predicción")
axes[0].set_ylabel("Valor Real")

cm_sk = confusion_matrix(y, preds_sk)
sns.heatmap(cm_sk, annot=True, fmt="d", cmap="Greens", ax=axes[1], cbar=False)
axes[1].set_title("Scikit-Learn (L-BFGS)")
axes[1].set_xlabel("Predicción")
axes[1].set_ylabel("Valor Real")

plt.tight_layout()
plt.show()

# -------------------------------------------------------------
# VISUALIZACIÓN 2: FRONTERAS DE DECISIÓN
# -------------------------------------------------------------
plt.figure(figsize=(10, 5))

sns.scatterplot(
    x=X[:, 0],
    y=X[:, 1],
    hue=y,
    palette={0: "red", 1: "green"},
    s=70,
    alpha=0.7,
)

x1_vals = np.linspace(X[:, 0].min(), X[:, 0].max(), 100)

x2_nr = -(pystat_nr.intercept_ + pystat_nr.coef_[0] * x1_vals) / pystat_nr.coef_[
    1
]
plt.plot(
    x1_vals, x2_nr, label="PyStat (Newton-Raphson)", color="orange", linewidth=2.5
)

x2_sk = -(
    sklearn_model.intercept_[0] + sklearn_model.coef_[0][0] * x1_vals
) / sklearn_model.coef_[0][1]
plt.plot(
    x1_vals,
    x2_sk,
    label="Scikit-Learn",
    color="black",
    linestyle="--",
    linewidth=2,
)

plt.title("Frontera de Decisión: PyStatistics vs. Scikit-Learn")
plt.xlabel("Horas de Estudio")
plt.ylabel("Puntaje Previo")
plt.legend()
plt.ylim(X[:, 1].min() - 5, X[:, 1].max() + 5)
plt.show()
```

```python
import time
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LogisticRegression as SklearnLogistic
from sklearn.metrics import accuracy_score, confusion_matrix, log_loss

print("=" * 60)
print("BENCHMARKING COMPLETO: PyStatistics (GD vs NR) vs. Scikit-Learn")
print("=" * 60)

# -------------------------------------------------------------
# 1. PyStatistics - Descenso del Gradiente
# -------------------------------------------------------------
start_time = time.time()
gd_opt = GradientDescent(learning_rate=0.01, max_iter=2000, tol=1e-6)
pystat_gd = LogisticRegression(optimizer=gd_opt)
pystat_gd.fit(X, y)
time_pystat_gd = time.time() - start_time

preds_gd = pystat_gd.predict(X)
probs_gd = pystat_gd.predict_proba(X)

# -------------------------------------------------------------
# 2. PyStatistics - Newton-Raphson
# -------------------------------------------------------------
start_time = time.time()
nr_opt = NewtonRaphson(max_iter=50, tol=1e-6)
pystat_nr = LogisticRegression(optimizer=nr_opt)
pystat_nr.fit(X, y)
time_pystat_nr = time.time() - start_time

preds_nr = pystat_nr.predict(X)
probs_nr = pystat_nr.predict_proba(X)

# -------------------------------------------------------------
# 3. Scikit-Learn (Sin regularización: C=np.inf)
# -------------------------------------------------------------
start_time = time.time()
sklearn_model = SklearnLogistic(C=np.inf, solver="lbfgs", max_iter=1000)
sklearn_model.fit(X, y)
time_sklearn = time.time() - start_time

preds_sk = sklearn_model.predict(X)
probs_sk = sklearn_model.predict_proba(X)[:, 1]

# -------------------------------------------------------------
# TABLA DE RESULTADOS COMPLETA
# -------------------------------------------------------------
results = pd.DataFrame(
    {
        "Modelo / Optimizador": [
            "PyStat (Gradient Descent)",
            "PyStat (Newton-Raphson)",
            "Scikit-Learn (L-BFGS)",
        ],
        "Iteraciones": [
            len(gd_opt.history),
            len(nr_opt.history),
            sklearn_model.n_iter_[0],
        ],
        "Tiempo (s)": [
            f"{time_pystat_gd:.4f}",
            f"{time_pystat_nr:.4f}",
            f"{time_sklearn:.4f}",
        ],
        "Accuracy": [
            accuracy_score(y, preds_gd),
            accuracy_score(y, preds_nr),
            accuracy_score(y, preds_sk),
        ],
        "Log-Loss": [
            log_loss(y, probs_gd),
            log_loss(y, probs_nr),
            log_loss(y, probs_sk),
        ],
        "Intercepto ($b$)": [
            f"{pystat_gd.intercept_:.4f}",
            f"{pystat_nr.intercept_:.4f}",
            f"{sklearn_model.intercept_[0]:.4f}",
        ],
        "Coef ($w_1, w_2$)": [
            np.round(pystat_gd.coef_, 4),
            np.round(pystat_nr.coef_, 4),
            np.round(sklearn_model.coef_[0], 4),
        ],
    }
)

display(results)

# -------------------------------------------------------------
# VISUALIZACIÓN: FRONTERAS DE DECISIÓN DE LOS 3 MODELOS
# -------------------------------------------------------------
plt.figure(figsize=(11, 6))

# Dibujar datos de entrenamiento
sns.scatterplot(
    x=X[:, 0],
    y=X[:, 1],
    hue=y,
    palette={0: "red", 1: "green"},
    s=70,
    alpha=0.7,
)

x1_vals = np.linspace(X[:, 0].min(), X[:, 0].max(), 100)

# 1. Frontera - Descenso del Gradiente (Incompleta/Lenta)
x2_gd = -(pystat_gd.intercept_ + pystat_gd.coef_[0] * x1_vals) / pystat_gd.coef_[
    1
]
plt.plot(
    x1_vals,
    x2_gd,
    label="PyStat (Gradient Descent - No convergió)",
    color="blue",
    linestyle=":",
    linewidth=2.5,
)

# 2. Frontera - Newton-Raphson
x2_nr = -(pystat_nr.intercept_ + pystat_nr.coef_[0] * x1_vals) / pystat_nr.coef_[
    1
]
plt.plot(
    x1_vals,
    x2_nr,
    label="PyStat (Newton-Raphson)",
    color="orange",
    linewidth=2.5,
)

# 3. Frontera - Scikit-Learn
x2_sk = -(
    sklearn_model.intercept_[0] + sklearn_model.coef_[0][0] * x1_vals
) / sklearn_model.coef_[0][1]
plt.plot(
    x1_vals,
    x2_sk,
    label="Scikit-Learn (L-BFGS)",
    color="black",
    linestyle="--",
    linewidth=2,
)

plt.title("Frontera de Decisión: GD vs. NR vs. Scikit-Learn")
plt.xlabel("Horas de Estudio")
plt.ylabel("Puntaje Previo")
plt.legend(loc="upper left")
plt.ylim(X[:, 1].min() - 5, X[:, 1].max() + 5)
plt.show()
```
